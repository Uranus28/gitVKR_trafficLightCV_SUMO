```python
def getMaxMinCort(a, b):
   return (a, b) if a > b else (b, a)
def getFarthestNearestDots(a, b, dots):
   return (dots[0], dots[1]) if a > b else (dots[1], dots[0])


def getRightLeftTopBottom(line1, line2):
   right, left = getFarthestNearestDots(line1[0][0], line1[1][0], line1)
   bottom, top = getFarthestNearestDots(line2[0][1], line2[1][1], line2)
   return right, left, top, bottom

def drawLine(img, coordinates, color=(128, 0, 32)):
   coord0_int = tuple(int(x) for x in coordinates[0])
   coord1_int = tuple(int(x) for x in coordinates[1])

   cv2.line(img, coord0_int, coord1_int, color, 2)
```

```python
def find_fourth_point(a, b, c):
    """
    a, b, c — кортежи или списки вида (x, y).
    Предполагается, что b — общая вершина для сторон ab и bc.
    """
    dx = a[0] + c[0] - b[0]
    dy = a[1] + c[1] - b[1]
    return (dx, dy)

# Пример использования:
A = (1, 1)
B = (4, 2)
C = (5, 5)

D = find_fourth_point(A, B, C)
print(f"Координаты 4-й точки: {D}")
```

```python
class DrawLineWidget(object):
    def __init__(self, img):
        self.original_image = img
        self.clone = self.original_image.copy()

        cv2.namedWindow('image')
        cv2.setMouseCallback('image', self.extract_coordinates)

        # List to store start/end points
        self.image_coordinates = []

    def extract_coordinates(self, event, x, y, flags, parameters):
        # Record starting (x,y) coordinates on left mouse button click
        if event == cv2.EVENT_LBUTTONDOWN:
            if (len(self.image_coordinates) < 2):
                self.image_coordinates = [(x,y)]

        # Record ending (x,y) coordinates on left mouse bottom release
        elif event == cv2.EVENT_LBUTTONUP:
            if (len(self.image_coordinates) < 2):
                self.image_coordinates.append((x,y))
                print('Starting: {}, Ending: {}'.format(self.image_coordinates[0], self.image_coordinates[1]))

                # Draw line
                drawLine(self.clone, self.image_coordinates)
                cv2.imshow("image", self.clone) 

        # Clear drawing boxes on right mouse button click
        elif event == cv2.EVENT_RBUTTONDOWN:
            self.clone = self.original_image.copy()
            self.image_coordinates = []

    def getImageCoordinates(self):
        return self.image_coordinates
    def show_image(self):
        return self.clone
```

```python
from enum import Enum
from itertools import chain

class Direction(str, Enum):
    '''
    сторона относительно линии
    '''
    LEFT = 'left'
    RIGHT = 'right'
def getReverseDirection(direc: Direction):
    '''
    обратное направление
    '''
    return Direction.LEFT if direc == Direction.RIGHT else Direction.RIGHT
class VerticalDirection(str, Enum):
    '''
    Вертикальное направление
    '''
    ToTop = 'ToTop'
    ToBottom = 'ToBottom'
class HorizontalDirection(str, Enum):
    '''
    Горизонтальное направление
    '''
    ToRight = 'ToRight'
    ToLeft = 'ToLeft'
# Все направления
AllDirections = Enum(
    'AllDirections',
    {item.name: item.value for item in chain(HorizontalDirection, VerticalDirection)}
)

class AreaCoords:
    def __init__(self, top_left=None, top_right=None, bottom_right=None, bottom_left=None):
        self.top_left = top_left
        self.top_right = top_right
        self.bottom_right = bottom_right
        self.bottom_left = bottom_left

    def set_coords(self, top_left=None, top_right=None, bottom_right=None, bottom_left=None):
        if top_left != None:
            self.top_left = top_left
        if top_right != None:
            self.top_right = top_right
        if bottom_right != None:
            self.bottom_right = bottom_right
        if bottom_left != None:
            self.bottom_left = bottom_left

    def is_any_none(self):
        '''
        Есть ли хоть одна координата None
        '''
        return True if self.top_left == None or self.top_right == None or self.bottom_left == None or self.bottom_right == None else False
    
    def get_area_coords(self):
        '''
        Returns array of coords
        '''
        return np.array([self.top_left, self.top_right, self.bottom_right, self.bottom_left])
    
    def get_bottom_coords(self, direction: HorizontalDirection):
        '''
        Returns array of bottom coords
        direction - direction to last dot in array
        '''
        return [self.bottom_right, self.bottom_left
             ] if direction == HorizontalDirection.ToLeft else [self.bottom_left, self.bottom_right]
    def get_right_coords(self, direction: VerticalDirection):
        '''
        Returns array of right coords
        direction - direction to last dot in array
        '''
        return [self.top_right, self.bottom_right
                ] if direction == VerticalDirection.ToBottom else [self.bottom_right, self.top_right]
    def get_top_coords(self, direction: HorizontalDirection):
        '''
        Returns array of top coords
        direction - direction to last dot in array
        '''
        return [self.top_left, self.top_right] if direction == HorizontalDirection.ToRight else [self.top_right, self.top_left]
    def get_left_coords(self, direction: VerticalDirection):
        '''
        Returns array of left coords
        direction - direction to last dot in array
        '''
        return [self.bottom_left, self.top_left] if direction == VerticalDirection.ToTop else [self.top_left, self.bottom_left]
    
    def state_dict(self):
      return {
        'top_left': self.top_left,
        'top_right': self.top_right,
        'bottom_left': self.bottom_left,
        'bottom_right': self.bottom_right,
      }

    def load_state_dict(self, state):
      self.top_left = tuple(state['top_left'])
      self.top_right = tuple(state['top_right'])
      self.bottom_left = tuple(state['bottom_left'])
      self.bottom_right = tuple(state['bottom_right'])
```

```python
from shapely import LineString

def isLineInsideCorrectArea(main_line: LineString, object_line: LineString, correct_position: Direction, withIntersect = True) -> str:
    """
    Проверяет, если крайняя граница объекта к встречному движению пересекает либо находится в нужном от него напралении
    (для движения спарва налево, проверяем нижнюю границу объекта)
    main_line - Линия направления движения (координаты передаются по направлению движения)
    object_line - крайняя линия объекта
    correct_position - требуемое расположение линии объекта относительно основной
    returns: boolean
    """
    if withIntersect and object_line.intersects(main_line):
        return True
    # Смещаем на 1 единицу влево и вправо
    left = main_line.offset_curve(1.0)
    right = main_line.offset_curve(-1.0)
    
    # Сравниваем расстояния
    position = Direction.LEFT if left.distance(object_line) < right.distance(object_line) else Direction.RIGHT
    return position == correct_position

# Пример
ref = LineString([(0, 0), (5, 0)])
print(isLineInsideCorrectArea(ref, LineString([(1, 1), (4, 1)]), Direction.LEFT))   # left
```

```python
import shapely
from shapely.geometry import LineString, Polygon

def getInfiniteRay(line:LineString):
    '''
    Создает луч из данного вектора
    '''
    a, b =  line.coords[0], line.coords[-1] 
    # a, b - кортежи (x, y)
    dx = b[0] - a[0]
    dy = b[1] - a[1]
    
    # т.к. луч - бесконечная прямая, 
    # умножаем вектор на очень большое число (100000) 
    # и строим отрезок от точки A
    scale = 100000
    p2_x = a[0] + dx * scale
    p2_y = a[1] + dy * scale
    
    return LineString([a, (p2_x, p2_y)])

def getIntersectionPoint(line1: LineString, line2: LineString):
    '''
    Получает точку пересечения двух линий, которые точно должны пересечься
    '''
    return line1.intersection(line2)
def getNextIntersectPoint(line1:LineString, line2:LineString):
    '''
    Получает на вход два вектора и возвращает точку, в которой они сомкнутся
    '''
    ray1 = getInfiniteRay(line1)
    ray2 = getInfiniteRay(line2)
    return getIntersectionPoint(ray1, ray2)
```

```python
def isIntersectOrInside(square_coords, area_coords):
    '''
    Проверяет есть ли общие точки у квадрата и области 
    (так-же проверяет, если крайняя граница объекта к встречному движению пересекает либо находится в нужном от него напралении)
    (для движения спарва налево, проверяем нижнюю границу объекта)
    square_coords - координаты квадрата
    area_coords - координаты области
    '''
    square = Polygon(square_coords)
    target_area = Polygon(area_coords)
    # 1. Проверить, пересекает ли квадрат область (есть ли общие точки)
    if square.intersects(target_area):
        return True
    # 2. Проверить, находится ли квадрат ПОЛНОСТЬЮ внутри области
    if square.within(target_area):
        return True
    return False
```

```python
import json
from collections import Counter
import random


class VideoConverterClassV2:

  def __init__(self, pathToVideoFolder = '/content/drive/MyDrive/2025Mag2/НИР/Видео/', videoImagesFolder = '/content/drive/MyDrive/2025Mag2/НИР/Видео/imagesFolder/', tracker_choice='botsort'):
    self.pathToVideoFolder = pathToVideoFolder # путь до папки с видео
    self.videoImagesFolder = videoImagesFolder # путь до папки, для сохранения отдельно предсказанных изображения изображений
    self.setTrackerChoice(tracker_choice)
    # количество кадров в секунду
    self.fps = 24
    self.info_color = (102,0,102)
    self.pred_color = (32,178,170)
    self.default_color = (128, 128, 128)
    # Координаты задающие направления на перекрестке
    self.area_coord_right_to_left = AreaCoords();
    self.crossroad_right_to_left_lines_coordinates = []
    self.right_to_left_color = (0,255,0)
    self.area_coord_top_to_bottom = AreaCoords();
    self.crossroad_top_to_bottom_lines_coordinates = []
    self.top_to_bottom_color = (255, 80, 255)
    self.area_coord_left_to_right = AreaCoords();
    self.crossroad_left_to_right_lines_coordinates = []
    self.left_to_right_color = (0, 168, 243)
    self.area_coord_bottom_to_top = AreaCoords();
    self.crossroad_bottom_to_top_lines_coordinates = []
    self.bottom_to_top_color = (255, 236, 20)
    # координаты области перекрестка
    self.area_crossroad = AreaCoords();
    self.area_crossroad_color = (127,255,212)
    self.arrayToDraw = np.array(['car', 'person', 'bus', 'truck', 'motorcycle'])
    self.windowNameForLine = 'image'
    self.json_lines_name = 'json_lines.json'
    self.csv_name = 'collected_info.csv'
    self.json_folder_name = '/json_lines'
    # Данные о дорожной ситуации
    '''
    'time_of_info' - время, когда получена информация
    'count_to...' - текущее количество автомобилей в направлении ...
    'count_stuck_middle' - текущее количесство автомобилей, застрявших посреди перекрестка
    'count_stuck_to...' - текущее количесство автомобилей, застрявших в направлении ...
    'is_jam_next...' - наличие пробки у следующего перекрестка по направлению ...
    *застрявших означает что объект находится в области больше одной итерации.*
    'is_green_horizontal' - В данный момент горит зеленый для горизонтальных направлений
    'is_green_vertical' - В данный момент горит зеленый для вертикальных направлений
    'y' - направление которому необходимо дать зеленый сигнал 
    '''
    self.columns = ['time_of_info',
                    'count_to_r',
                    'count_to_l',
                    'count_to_b',
                    'count_to_t',
                    'count_stuck_middle',
                    'count_stuck_to_r',
                    'count_stuck_to_l',
                    'count_stuck_to_b',
                    'count_stuck_to_t',
                    'is_jam_next_l',
                    'is_jam_next_r',
                    'is_jam_next_t',
                    'is_jam_next_b',
                    'is_green_horizontal',
                    'is_green_vertical',
                    'y',
                    ]
    self.feature_cols = self.columns[1:-1]
    self.pred_col = self.columns[-1]
    # датафрейм для сохранения информации о ситуации на перекрестке
    self.trafficInfoDF = pd.DataFrame(columns=self.columns)
    # счетчик итераций для машин, застрявших в указанной области
    self.objects_stuck_middle = Counter()
    self.objects_stuck_to_r = Counter()
    self.objects_stuck_to_l = Counter()
    self.objects_stuck_to_b = Counter()
    self.objects_stuck_to_t = Counter()
    # сеты машин, в указанной области
    self.objects_to_r = set()
    self.objects_to_l = set()
    self.objects_to_b = set()
    self.objects_to_t = set()

  def save_lines_to_json(self, folder_to_save):
    '''
    Сохраняет данные об областях в json файлик
    json файл сохраняется в папку self.json_folder_name в папке, где хранятся кадры видео
    '''
    data = {
      'area_coord_right_to_left': self.area_coord_right_to_left.state_dict(),
      'area_coord_top_to_bottom': self.area_coord_top_to_bottom.state_dict(),
      'area_coord_left_to_right': self.area_coord_left_to_right.state_dict(),
      'area_coord_bottom_to_top': self.area_coord_bottom_to_top.state_dict(),
      'area_crossroad': self.area_crossroad.state_dict()
    }
    path_to_save_folder = folder_to_save + self.json_folder_name
    if not os.path.exists(path_to_save_folder):
      os.makedirs(path_to_save_folder)
    
    file_name = path_to_save_folder +'/'+ self.json_lines_name
    # Open file in write mode ('w')
    with open(file_name, "w") as file:
        json.dump(data, file)

  def save_dataframe(self, folder_to_save):
    '''
    Сохраняет собранные данные об областях в csv файлик
    csv файл сохраняется в папку self.json_folder_name в папке, где хранятся кадры видео
    '''

    path_to_save_folder = folder_to_save + self.json_folder_name
    fileListLen = str(len(glob(path_to_save_folder+'/*')))
    if not os.path.exists(path_to_save_folder):
      os.makedirs(path_to_save_folder)
    
    file_name = path_to_save_folder +'/'+fileListLen+'_'+ self.csv_name
    # Open file in write mode ('w')
    # сохранение собранной информации в csv
    self.trafficInfoDF.to_csv(file_name, index=False)

  def load_lines_from_json(self, folder_to_save):
    '''
    Загружает данные об областях из json файлика
    json файл должен находиться в папке self.json_folder_name в папке, где хранятся кадры видео
    '''
    path_to_save_folder = folder_to_save+self.json_folder_name
    file_name = path_to_save_folder +'/'+ self.json_lines_name
    if not os.path.exists(file_name):
      print("File with lines not found :(. Please draw new ones")
      return False
    else:
      try:
        # Open the file and parse its contents
        with open(file_name, 'r', encoding='utf-8') as file:
            data = json.load(file)
            self.area_coord_right_to_left.load_state_dict(data["area_coord_right_to_left"])
            self.area_coord_top_to_bottom.load_state_dict(data["area_coord_top_to_bottom"])
            self.area_coord_left_to_right.load_state_dict(data["area_coord_left_to_right"])
            self.area_coord_bottom_to_top.load_state_dict(data["area_coord_bottom_to_top"])
            self.area_crossroad.load_state_dict(data["area_crossroad"])
            print("lines loaded")
            return True
      except:
        print("Error occured while reading :(. Please draw new ones")
        return False

  def setTrackerChoice(self, tracker_choice):
    if tracker_choice not in ['botsort', 'bytetrack']:
        print(f"Invalid tracker choice '{tracker_choice}', defaulting to 'botsort'")
        self.tracker_choice = 'botsort'
    else:
       self.tracker_choice = tracker_choice

  def convertToVid(self, folder_name, video_name = 'vidSeg.avi'):
    '''
    Конвертация изображений в видео
    folder_name - Название папки с сегментированными изображениями
    video_name - название собранного видео
    '''
    pathToVideoImages = self.videoImagesFolder + folder_name
    if not os.path.exists(pathToVideoImages):
      print("Try run Mask Video first")
    else:

      images = [img for img in os.listdir(pathToVideoImages) if img.endswith((".jpg", ".jpeg", ".png"))]
      print("Images:", images)

      # Set frame from the first image
      frame = cv2.imread(os.path.join(pathToVideoImages, images[0]))
      height, width, layers = frame.shape

      # Video writer to create .avi file
      pathToVideo = self.pathToVideoFolder + video_name
      video = cv2.VideoWriter(pathToVideo, cv2.VideoWriter_fourcc(*'DIVX'), self.fps, (width, height))

      # Appending images to video
      for image in images:
          video.write(cv2.imread(os.path.join(pathToVideoImages, image)))

      # Release the video file
      video.release()
      cv2.destroyAllWindows()
      print("Video generated successfully!")

  def predictSingleImg(self, img, model, pathToVideoImages, curIm):
    '''
    Predict single img with the model
    img - Изображение для сегментации
    model - обученная модель для сегментации
    '''
    results = model(img)  # predict on an image

    # Access the results
    for result in results:
      name = pathToVideoImages + "/" + str(curIm).zfill(32) + '.jpg'
      result.save(filename=name)

    return result
  
  def addLineOnImage(self, img, label):
     cv2.putText(img, f"draw line: {label}", (10, 70),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, self.info_color, 3)
     cv2.putText(img, f"q - skip draw \n e - continue \n rClick - clear line", (10, 150),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, self.info_color, 3)
     draw_line_widget = DrawLineWidget(img)
     while True:
      cv2.imshow(self.windowNameForLine, draw_line_widget.show_image())

      key = cv2.waitKey(1) & 0xFF

      # Close program with keyboard 'q'
      if key == ord('q'):
          cv2.destroyWindow(self.windowNameForLine)
          return []
      if key == ord('e'):
        line_coords = draw_line_widget.getImageCoordinates()
        print(line_coords)
        if (len(line_coords) != 2):
          cv2.putText(img, f"add line", (10, 230),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 3)
        else:
          cv2.destroyWindow(self.windowNameForLine)
          return line_coords
  
  def drawAllCoords(self, img, area: AreaCoords, color=(128, 0, 32)):
      drawLine(img, area.get_bottom_coords(HorizontalDirection.ToLeft), color)
      drawLine(img, area.get_right_coords(VerticalDirection.ToBottom), color)
      drawLine(img, area.get_top_coords(HorizontalDirection.ToLeft), color)
      drawLine(img, area.get_left_coords(VerticalDirection.ToBottom), color)

  def drawSeparatorLines(self, img):
    if self.area_coord_right_to_left.is_any_none() == False:
      self.drawAllCoords(img, self.area_coord_right_to_left)
    if self.area_coord_bottom_to_top.is_any_none() == False:
      self.drawAllCoords(img, self.area_coord_bottom_to_top)
    if self.area_coord_left_to_right.is_any_none() == False:
      self.drawAllCoords(img, self.area_coord_left_to_right)
    if self.area_coord_top_to_bottom.is_any_none() == False:
      self.drawAllCoords(img, self.area_coord_top_to_bottom)
    if self.area_crossroad.is_any_none() == False:
      self.drawAllCoords(img, self.area_crossroad, self.area_crossroad_color)
  
  def isCorrectLines(self, main_line, border_line, object_line, correct_position: Direction):
    # т.к. в видео захвате y сверх вниз делаем реверс correct_position
    main_correct_position = getReverseDirection(correct_position)
    border_correct_position = Direction.LEFT if main_correct_position == Direction.RIGHT else Direction.RIGHT
    mainLineString = LineString(main_line)
    objectLineString = LineString(object_line)
    borderLineString = LineString(border_line)
    return isLineInsideCorrectArea(
          main_line=mainLineString,
          object_line=objectLineString,
          correct_position=main_correct_position) and isLineInsideCorrectArea(
          main_line=borderLineString,
          object_line=objectLineString,
          correct_position=border_correct_position,
          withIntersect=False)
    

  def checkIfObjectInsideCorrectArea(self, box_coords, objectBox: AreaCoords, areaDirection: HorizontalDirection | VerticalDirection):
    '''
    В случае, когда объект касается нужной области проверяет,
    если этот объект находится на корректной стороне движения и не на обочине
    '''
    match areaDirection:
      case HorizontalDirection.ToLeft:
        areaObject = self.area_coord_right_to_left
        # В случае, если объект касается области проверяем, если 
        if isIntersectOrInside(box_coords, areaObject.get_area_coords()) == False:
          return False
        else:
          object_line = objectBox.get_bottom_coords(HorizontalDirection.ToLeft)
          # print(areaObject.get_area_coords(), object_line)
          return self.isCorrectLines(
            main_line=areaObject.get_bottom_coords(HorizontalDirection.ToLeft),
            object_line=object_line,
            border_line=areaObject.get_top_coords(HorizontalDirection.ToLeft),
            correct_position=Direction.RIGHT
          )
      case HorizontalDirection.ToRight:
        areaObject = self.area_coord_left_to_right
        # В случае, если объект касается области проверяем, если 
        if isIntersectOrInside(box_coords, areaObject.get_area_coords()) == False:
          return False
        else:
          object_line = objectBox.get_top_coords(HorizontalDirection.ToRight)
          return self.isCorrectLines(
            main_line=areaObject.get_top_coords(HorizontalDirection.ToRight),
            object_line=object_line,
            border_line=areaObject.get_bottom_coords(HorizontalDirection.ToRight),
            correct_position=Direction.RIGHT
          )
      case VerticalDirection.ToBottom:
        areaObject = self.area_coord_top_to_bottom
        # В случае, если объект касается области проверяем, если 
        if isIntersectOrInside(box_coords, areaObject.get_area_coords()) == False:
          return False
        else:
          object_line = objectBox.get_right_coords(VerticalDirection.ToBottom)
          return self.isCorrectLines(
            main_line=areaObject.get_right_coords(VerticalDirection.ToBottom),
            object_line=object_line,
            border_line=areaObject.get_left_coords(VerticalDirection.ToBottom),
            correct_position=Direction.RIGHT
          )
      case VerticalDirection.ToTop:
        areaObject = self.area_coord_bottom_to_top
        # В случае, если объект касается области проверяем, если 
        if isIntersectOrInside(box_coords, areaObject.get_area_coords()) == False:
          return False
        else:
          object_line = objectBox.get_left_coords(VerticalDirection.ToTop)
          return self.isCorrectLines(
            main_line=areaObject.get_left_coords(VerticalDirection.ToTop),
            object_line=object_line,
            border_line=areaObject.get_right_coords(VerticalDirection.ToTop),
            correct_position=Direction.RIGHT
          )
      case _:
        return False
      
  def getTrullyStuck(self, _counter: Counter):
    '''
    Возвращает количество машин, которые простояли в области n больше 2 итераций
    '''
    return len([item for item, count in _counter.items() if count > 2])
  
  def addToDirectionSet(self, _set: set, trackId):
    if trackId not in _set:
      _set.add(trackId)
  def addToDirectionCounter(self, _counter: Counter, trackId, skipCheck=False):
    if skipCheck or trackId not in _counter:
      _counter.update([trackId])
    
  def removeFromSet(self, _set: set, trackId):
    if trackId in _set:
      _set.remove(trackId)
  def removeFromCounter(self, _counter: Counter, trackId):
    if trackId in _counter:
      del _counter[trackId]

  def removeFromOtherStuckCounters(self, trackId):
    self.removeFromCounter(self.objects_stuck_to_b, trackId)
    self.removeFromCounter(self.objects_stuck_to_l, trackId)
    self.removeFromCounter(self.objects_stuck_to_r, trackId)
    self.removeFromCounter(self.objects_stuck_to_t, trackId)

  def removeFromOtherCurrentSets(self, trackId):
    self.removeFromSet(self.objects_to_b, trackId)
    self.removeFromSet(self.objects_to_l, trackId)
    self.removeFromSet(self.objects_to_r, trackId)
    self.removeFromSet(self.objects_to_t, trackId)
  
  def addToMiddleCounter(self, trackId):
    if trackId not in self.objects_stuck_middle:
      self.objects_stuck_middle.update([trackId])
      self.removeFromOtherStuckCounters(trackId)
      self.removeFromOtherCurrentSets(trackId)

  def getLinesOnFrame(self, frame_result, folder_to_save):
    '''
    Получает отрисованные пользователем линии для областей наблюдения
    '''
    # линии для справа налево
    line1 = self.addLineOnImage(frame_result.orig_img.copy(), "separating traffic right to left with pedestrian crossing")
    if (len(line1)):
      self.crossroad_right_to_left_lines_coordinates.append(line1)
    line2 = self.addLineOnImage(frame_result.orig_img.copy(), "pedestrian crossing on right to left")
    if (len(line2)):
      self.crossroad_right_to_left_lines_coordinates.append(line2)
    if (len(line2) and len(line1)):
      right, left, top, bottom = getRightLeftTopBottom(line1, line2)
      #!!!!!!!! нужно получить угол по которому стоит дорога
      coord_border = find_fourth_point(right, left, top)
      self.crossroad_right_to_left_lines_coordinates.append([top, coord_border])
      self.area_coord_right_to_left.set_coords(
        top_left=top,
        top_right=coord_border, 
        bottom_left=left, 
        bottom_right=right
      )
    else:
      self.crossroad_right_to_left_lines_coordinates = []

    # линии для сверху вниз
    line1 = self.addLineOnImage(frame_result.orig_img.copy(), "separating traffic top to bottom with pedestrian crossing")
    if (len(line1)):
      self.crossroad_top_to_bottom_lines_coordinates.append(line1)
    line2 = self.addLineOnImage(frame_result.orig_img.copy(), "pedestrian crossing on top to bottom")
    if (len(line2)):
      self.crossroad_top_to_bottom_lines_coordinates.append(line2)
    if (len(line2) and len(line1)):
      right, left, top, bottom = getRightLeftTopBottom(line2, line1)
      coord_border = find_fourth_point(left, bottom, top)
      self.area_coord_top_to_bottom.set_coords(
        top_left=coord_border, 
        top_right=top, 
        bottom_left=left, 
        bottom_right=bottom
      )
      self.crossroad_top_to_bottom_lines_coordinates.append([left, coord_border])
    else:
      self.crossroad_top_to_bottom_lines_coordinates = []
    
    # линии для слева направо
    line1 = self.addLineOnImage(frame_result.orig_img.copy(), "separating traffic left to right with pedestrian crossing")
    if (len(line1)):
      self.crossroad_left_to_right_lines_coordinates.append(line1)
    line2 = self.addLineOnImage(frame_result.orig_img.copy(), "pedestrian crossing on left to right")
    if (len(line2)):
      self.crossroad_left_to_right_lines_coordinates.append(line2)
    if (len(line2) and len(line1)):
      right, left, top, bottom = getRightLeftTopBottom(line1, line2)
      coord_border = find_fourth_point(left, right, bottom)
      self.area_coord_left_to_right.set_coords(
        top_left=left, 
        top_right=right, 
        bottom_left=coord_border, 
        bottom_right=bottom
      )
      self.crossroad_left_to_right_lines_coordinates.append([bottom, coord_border])
    else:
      self.crossroad_left_to_right_lines_coordinates = []
    
    # линии для снизу вверх
    line1 = self.addLineOnImage(frame_result.orig_img.copy(), "separating traffic bottom to top with pedestrian crossing")
    if (len(line1)):
      self.crossroad_bottom_to_top_lines_coordinates.append(line1)
    line2 = self.addLineOnImage(frame_result.orig_img.copy(), "pedestrian crossing on bottom to top")
    if (len(line2)):
      self.crossroad_bottom_to_top_lines_coordinates.append(line2)
    if (len(line2) and len(line1)):
      right, left, top, bottom = getRightLeftTopBottom(line2, line1)
      coord_border = find_fourth_point(bottom, top, right)
      self.area_coord_bottom_to_top.set_coords(
        top_left=top, 
        top_right=right, 
        bottom_left=bottom, 
        bottom_right=coord_border
      )
      self.crossroad_bottom_to_top_lines_coordinates.append([right, coord_border])
    else:
      self.crossroad_bottom_to_top_lines_coordinates = []

    if self.area_coord_bottom_to_top.is_any_none() == False and self.area_coord_left_to_right.is_any_none() == False and self.area_coord_top_to_bottom.is_any_none() ==False and self.area_coord_right_to_left.is_any_none() ==False:
      crossroad_top_right = getNextIntersectPoint(
        LineString(self.area_coord_right_to_left.get_left_coords(VerticalDirection.ToTop)),
        LineString(self.area_coord_top_to_bottom.get_bottom_coords(HorizontalDirection.ToRight))
                    )
      crossroad_top_left = getNextIntersectPoint(
        LineString(self.area_coord_left_to_right.get_right_coords(VerticalDirection.ToTop)),
        LineString(self.area_coord_top_to_bottom.get_bottom_coords(HorizontalDirection.ToLeft))
                    )
      crossroad_bottom_left = getNextIntersectPoint(
        LineString(self.area_coord_left_to_right.get_right_coords(VerticalDirection.ToBottom)),
        LineString(self.area_coord_bottom_to_top.get_top_coords(HorizontalDirection.ToLeft))
                    )
      crossroad_bottom_right = getNextIntersectPoint(
        LineString(self.area_coord_right_to_left.get_left_coords(VerticalDirection.ToBottom)),
        LineString(self.area_coord_bottom_to_top.get_top_coords(HorizontalDirection.ToRight))
                    )
      self.area_crossroad.set_coords(
          top_left=crossroad_top_left.coords[0], 
          top_right=crossroad_top_right.coords[0], 
          bottom_left=crossroad_bottom_left.coords[0], 
          bottom_right=crossroad_bottom_right.coords[0], 
        )
      self.save_lines_to_json(folder_to_save=folder_to_save)
  
  def getGreenLightModel(self, fileWithModelPath):
    if not os.path.exists(fileWithModelPath):
      print("Model NotFound")
      return None
    else:
      loaded_model = CatBoostClassifier()
      loaded_model.load_model("catboost_model.cbm")
      return loaded_model

  def maskVideoToImgs(self, fileName, folder_name, trained_model, second_to_switch = 5, showVideo=True, fileWithModelPath = 'catboost_model.cbm'):
    '''
    Сегментировать видео с сохранением изображений в отдельную папку
    fileName - путь до видео для сегментации
    folder_name - Название папки Для сохранения изображений
    trined_model - обученная модель для сегментации
    showVideo - показать обработанное видео в окне
    '''

    pathToVideoImages = self.videoImagesFolder + folder_name
    if not os.path.exists(pathToVideoImages):
      os.makedirs(pathToVideoImages)
    else:
      fileList = glob(pathToVideoImages+'/*')
      print("Number of files: ",len(fileList))
      for filePath in fileList:
        try:
            os.remove(filePath)
        except:
            print("Error while deleting file : ", filePath)

    print("обработанные кадры сохранятся в папку", pathToVideoImages, ". Если произошла ошибка, запустите makeVideo, чтобы собрать собранные кадры в видео")
    model = YOLO(str(trained_model.save_dir)+'/weights/best.pt')

    # Load the video file
    cap = cv2.VideoCapture(fileName)
    if not cap.isOpened():
        print(f"Error opening video file {fileName}")
        return
 
    # Initialize variables for FPS calculation
    prev_time = 0
 
    """
    Parameters of track method:
    source: str - path to video file or camera index
    tracker: str - tracker configuration file (e.g., 'bytetrack.yaml' or 'botsort.yaml')
    conf: float - confidence threshold for detections
    stream: bool - if True, yields frames one by one for real-time processing, 
                if False, processes the entire video at once
    """

    model_to_predict = self.getGreenLightModel(fileWithModelPath=fileWithModelPath)
    is_model_to_predict = model_to_predict is not None
    results = model.track(
        source=fileName, # path to video file 
        tracker=f'{self.tracker_choice}.yaml',  # 'bytetrack.yaml' or 'botsort.yaml'
        conf=0.25,   # confidence threshold                      
        stream=True # set it to True for continuous video processing                    
    )
    print(self.tracker_choice)
    curIm = 0
    fpsAvg = 0
    # loop through the results
    for frame_result in results:
        isSwitchFrame = curIm % 100 == 0
        isSwitchFrameX2 = curIm % 200 == 0
        isWriteInfoFrame = curIm % 50 == 0
        count_right_left = 0
        count_top_bottom = 0
        count_left_right = 0
        count_bottom_top = 0
        count_crossroad = 0
        if curIm == 0:
          if self.load_lines_from_json(folder_to_save=pathToVideoImages) == False:
            self.getLinesOnFrame(frame_result, pathToVideoImages)
        # print(frame_result)
        # Get the original frame
        img = frame_result.orig_img.copy()
        masks = []
        if (masks in frame_result):
          masks = frame_result.masks.data.cpu().numpy() 
 
        # Calculate FPS (fall back to input_fps until prev_time set)
        curr_time = time.time()
        fps = 1 / (curr_time - prev_time) if prev_time != 0 else cap.get(cv2.CAP_PROP_FPS)
        prev_time = curr_time

        # draw crossroad lines
        if (showVideo):
          self.drawSeparatorLines(img)
        # loop through the detected boxes and draw them on the frame
        for i, box in enumerate(frame_result.boxes):
            # Extract class
            cls = int(box.cls.cpu().item())
            """
            model.names is a dictionary mapping class indices to class names.
            {0: 'person',
            1: 'bicycle',
            2: 'car',
            3: 'motorcycle',
            ...,
            }
            """
            class_name = model.names[cls]
            # учитываем только необходимые
            if (np.any(np.array(self.arrayToDraw == class_name))):
              # Extract bounding box coordinates
              x1, y1, x2, y2 = map(int, box.xyxy.cpu().numpy()[0])
              # Extract confidence
              conf = box.conf.cpu().item()
              # Extract track ID 
              track_id = int(box.id.cpu().item()) if box.id is not None else -1

              if (showVideo and len(masks)):
                mask = masks[i]
                # Resize mask to original image size (often necessary if model resized input)
                mask_resized = cv2.resize(mask, (img.shape[1], img.shape[0]))
                
                # 3. Create a colored overlay
                color = [0, 255, 0]  # Green mask
                colored_mask = np.zeros_like(img, dtype=np.uint8)
                colored_mask[mask_resized > 0.5] = color
              
                # 4. Apply to image with transparency (0.5 weight)
                img = cv2.addWeighted(img, 1.0, colored_mask, 0.5, 0)
              box_top_left = (x1, y1)
              box_top_right = (x2, y1)
              box_bottom_right = (x2, y2)
              box_bottom_left = (x1, y2)
              box_area = AreaCoords(bottom_left=box_bottom_left, bottom_right=box_bottom_right,
                                    top_left=box_top_left, top_right=box_top_right)

              isColored = False
              
              box_coords = box_area.get_area_coords()
              # Цвет для закраски границы
              colorToBox = self.default_color
              if self.area_coord_right_to_left.is_any_none() == False:
                if self.checkIfObjectInsideCorrectArea(
                  box_coords=box_coords,
                  objectBox=box_area,
                  areaDirection=HorizontalDirection.ToLeft
                ):
                  self.addToDirectionSet(_set=self.objects_to_l, trackId=track_id)
                  self.addToDirectionCounter(_counter=self.objects_stuck_to_l, trackId=track_id, skipCheck=isSwitchFrame)
                  colorToBox = self.right_to_left_color
                  count_right_left += 1
                  isColored = True
              if isColored == False and self.area_coord_top_to_bottom.is_any_none() == False:
                if self.checkIfObjectInsideCorrectArea(
                  box_coords=box_coords,
                  objectBox=box_area,
                  areaDirection=VerticalDirection.ToBottom
                ):
                  self.addToDirectionSet(_set=self.objects_to_b, trackId=track_id)
                  self.addToDirectionCounter(_counter=self.objects_stuck_to_b, trackId=track_id, skipCheck=isSwitchFrame)
                  colorToBox = self.top_to_bottom_color
                  count_top_bottom += 1
                  isColored = True
              if isColored == False and self.area_coord_left_to_right.is_any_none() == False:
                if self.checkIfObjectInsideCorrectArea(
                  box_coords=box_coords,
                  objectBox=box_area,
                  areaDirection=HorizontalDirection.ToRight
                ):
                  self.addToDirectionSet(_set=self.objects_to_r, trackId=track_id)
                  self.addToDirectionCounter(_counter=self.objects_stuck_to_r, trackId=track_id, skipCheck=isSwitchFrame)
                  colorToBox = self.left_to_right_color
                  count_left_right += 1
                  isColored = True
              if isColored == False and self.area_coord_bottom_to_top.is_any_none() == False:
                if self.checkIfObjectInsideCorrectArea(
                  box_coords=box_coords,
                  objectBox=box_area,
                  areaDirection=VerticalDirection.ToTop
                ):
                  self.addToDirectionSet(_set=self.objects_to_t, trackId=track_id)
                  self.addToDirectionCounter(_counter=self.objects_stuck_to_t, trackId=track_id, skipCheck=isSwitchFrame)
                  count_bottom_top += 1
                  colorToBox = self.bottom_to_top_color
                  isColored = True
              if isColored == False and self.area_crossroad.is_any_none() == False:
                areaObject = self.area_crossroad
                # В случае, если объект касается области или находится внутри ее проверяем 
                if isIntersectOrInside(box_coords, areaObject.get_area_coords()):
                  self.addToMiddleCounter(trackId=track_id) if isSwitchFrame == False else self.addToDirectionCounter(_counter=self.objects_stuck_middle, trackId=track_id, skipCheck=isSwitchFrame)
                  count_crossroad += 1
                  colorToBox = self.area_crossroad_color
                  isColored = True
              if isColored == False:
                # возможно объект выехал из зоны наблюдения
                self.removeFromCounter(self.objects_stuck_middle, trackId=track_id)
              if (showVideo):  
                # Draw bounding box and label
                cv2.rectangle(img, (x1, y1), (x2, y2), colorToBox, 2)
                label = f"{class_name} ID:{track_id} {conf:.2f}"
                cv2.putText(img, label, (x1, y1 - 10),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.5, colorToBox, 2)
        if (isWriteInfoFrame):
          # Пока что рандомные числа 0 или 1 для следующего перекрестка
          # info_collected = {'time_of_info': curIm,
          #           'count_to_r': count_left_right,
          #           'count_to_l': count_right_left,
          #           'count_to_b': count_top_bottom,
          #           'count_to_t': count_bottom_top,
          #           'count_stuck_middle': self.getTrullyStuck(_counter=self.objects_stuck_middle),
          #           'count_stuck_to_r': self.getTrullyStuck(_counter=self.objects_stuck_to_r),
          #           'count_stuck_to_l': self.getTrullyStuck(_counter=self.objects_stuck_to_l),
          #           'count_stuck_to_b': self.getTrullyStuck(_counter=self.objects_stuck_to_b),
          #           'count_stuck_to_t': self.getTrullyStuck(_counter=self.objects_stuck_to_t),
          #           'is_jam_next_l': random.randint(0, 1),
          #           'is_jam_next_r': random.randint(0, 1),
          #           'is_jam_next_t': random.randint(0, 1),
          #           'is_jam_next_b': random.randint(0, 1),
          #           'is_green_horizontal': 1 if isSwitchFrameX2 else 0,
          #           'is_green_vertical': 0 if isSwitchFrameX2 else 1,
          #           'y': 1,
          # }
          # записываем информацию о ситуации на дороге в dataFrame
          info_collected = {'time_of_info': curIm,
                    'count_to_r': count_left_right,
                    'count_to_l': count_right_left,
                    'count_to_b': count_top_bottom,
                    'count_to_t': count_bottom_top,
                    'count_stuck_middle': self.getTrullyStuck(_counter=self.objects_stuck_middle),
                    'count_stuck_to_r': self.getTrullyStuck(_counter=self.objects_stuck_to_r),
                    'count_stuck_to_l': self.getTrullyStuck(_counter=self.objects_stuck_to_l),
                    'count_stuck_to_b': self.getTrullyStuck(_counter=self.objects_stuck_to_b),
                    'count_stuck_to_t': self.getTrullyStuck(_counter=self.objects_stuck_to_t),
                    'is_jam_next_l': 0,
                    'is_jam_next_r': 0,
                    'is_jam_next_t': 0,
                    'is_jam_next_b': 0,
                    'is_green_horizontal': 1 if isSwitchFrameX2 else 0,
                    'is_green_vertical': 0 if isSwitchFrameX2 else 1,
                    'y': 1,
          }
          dfInfo = pd.DataFrame(info_collected, index=[0])
          self.trafficInfoDF = pd.concat([self.trafficInfoDF, dfInfo], ignore_index=True)
          currentDirectionGreen = None
          if is_model_to_predict:
            X_to_pred = dfInfo[self.feature_cols]
            # y_to_pred = dfInfo[self.pred_col]  # 0, 1, 2
            pred = model_to_predict.predict(X_to_pred)[0]
            # print(pred)
            currentDirectionGreen = mapOfDirections[str(pred[0])]

        if (showVideo):           
          # Display tracker name and FPS on top-left corner
          cv2.putText(img, f"Tracker: {self.tracker_choice}", (10, 70),
                      cv2.FONT_HERSHEY_SIMPLEX, 2, self.info_color, 3)
          cv2.putText(img, f"FPS: {fps:.2f}", (10, 150),
                      cv2.FONT_HERSHEY_SIMPLEX, 2, self.info_color, 3)
          cv2.putText(img,
                      f"r-l: {count_right_left:.0f} | t-b: {count_top_bottom:.0f} | l-r: {count_left_right:.0f} | b-t: {count_bottom_top:.0f} | cross: {count_crossroad:.0f}", (10, 250),
                      cv2.FONT_HERSHEY_SIMPLEX, 1, self.info_color, 3)
          if currentDirectionGreen is not None:
            cv2.putText(img, f"switch to: {currentDirectionGreen}", (10, 350),
                      cv2.FONT_HERSHEY_SIMPLEX, 2, self.pred_color, 3)
          # Display the frame with detections  
          cv2.imshow('YOLOv26 Tracking', img)
          name_to_save = pathToVideoImages + "/" + str(curIm).zfill(32) + '.jpg'
          cv2.imwrite(name_to_save, img)
        # else:
        #   # Иначе показываем fps
        #   print(f"FPS: {fps:.2f} || r-l: {count_right_left:.0f} | t-b: {count_top_bottom:.0f} | l-r: {count_left_right:.0f} | b-t: {count_bottom_top:.0f}")
        if (curIm % 50 == 0):
          print("counters")
          print("mid", self.objects_stuck_middle)
          print("to_left", self.objects_stuck_to_l)
          print("to_right", self.objects_stuck_to_r)
          print("to_top", self.objects_stuck_to_t)
          print("to_bottom", self.objects_stuck_to_b)
          print("sets")
          print("to_left", self.objects_to_l)
          print("to_right", self.objects_to_r)
          print("to_top", self.objects_to_t)
          print("to_bottom", self.objects_to_b)
        curIm += 1
        fpsAvg += fps
        if curIm % 24 == 0:
          print(curIm)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print('avg FPS: ', fpsAvg/curIm)
            break
    
    # сохранение собранной информации в csv
    self.save_dataframe(folder_to_save=pathToVideoImages)

    print('avg FPS: ', fpsAvg/curIm)
    cap.release()
    cv2.destroyAllWindows()
```

```python
# Анализ ситуации на дороге
def get_Y(df: pd.DataFrame):
  '''
    столбцы 
    'time_of_info' - время, когда получена информация
    'count_to...' - текущее количество автомобилей в направлении ...
    'count_stuck_middle' - текущее количесство автомобилей, застрявших посреди перекрестка
    'count_stuck_to...' - текущее количесство автомобилей, застрявших в направлении ...
    'is_jam_next...' - наличие пробки у следующего перекрестка по направлению ...
    *застрявших означает что объект находится в области больше одной итерации.*
    'is_green_horizontal' - В данный момент горит зеленый для горизонтальных направлений
    'is_green_vertical' - В данный момент горит зеленый для вертикальных направлений
    'y' - направление которому необходимо дать зеленый сигнал 
  '''
  # делаем копию датасета
  df_c = df.copy()
  # Если центр перекрестка заблокирован — красный для всех
  stuckMiddleCondition = df['count_stuck_middle'] > 0
  df_c['cars_count'] = df[['count_to_r', 'count_to_l', 'count_to_b', 'count_to_t']].sum(axis=1)
  
  # Считаем Полезности включенного сигнала для каждого из направлений
  # доля от всех машин в каждом направлении + 
  #  + Количество машин которые стоят несколько фаз в каждом из направлений
  # (Для таких машин приоритет выше)

  # В случае, если по направлению вправо нету пробки
  df_c['utility_r'] = np.where(df_c['is_jam_next_r'] == 0,
                             (df_c['count_to_r'] / df_c['cars_count']) + df_c['count_stuck_to_r']
                             , 0)
  df_c['utility_l'] = np.where(df_c['is_jam_next_l'] == 0,
                             (df_c['count_to_l'] / df_c['cars_count']) + df_c['count_stuck_to_l']
                             , 0)
  df_c['utility_t'] = np.where(df_c['is_jam_next_t'] == 0,
                             (df_c['count_to_t'] / df_c['cars_count']) + df_c['count_stuck_to_t']
                             , 0)
  df_c['utility_b'] = np.where(df_c['is_jam_next_b'] == 0,
                             (df_c['count_to_b'] / df_c['cars_count']) + df_c['count_stuck_to_b']
                             , 0)
  
  # print(df_c.head())
  # решаем на основе полезности фаз с учётом пробок на следующих светофорах, какому направлению давать сигнал
  utilityComaprison = (df_c['utility_r'] + df_c['utility_l']) > (df_c['utility_t'] + df_c['utility_b'])

  conditions = [
    (stuckMiddleCondition == True),
    (utilityComaprison == True),
    (utilityComaprison != True)
  ]
  # choices = [DirectionOfGreen.NONE, DirectionOfGreen.HORIZONTAL, DirectionOfGreen.VERTICAL]
  choices = [0, 2, 1]
  # temp_series = np.select(conditions, choices, default=0)
  # return pd.Series(temp_series).map({0: DirectionOfGreen.NONE, 2: DirectionOfGreen.HORIZONTAL, 1:DirectionOfGreen.VERTICAL})
  return np.select(conditions, choices, default=0)
```

```python
import os
import cv2

def video_to_images(video_path, output_folder="extracted_frames"):
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    else:
        fileList = glob(output_folder+'/*')
        print("Number of files: ",len(fileList))
        for filePath in fileList:
            try:
                os.remove(filePath)
            except:
                print("Error while deleting file : ", filePath)
    # Initialize video capture object
    cap = cv2.VideoCapture(video_path)

    # Check if video opened successfully
    if not cap.isOpened():
        print(f"Error: Could not open video file {video_path}")
        return

    frame_count = 0

    while True:
        
        # Read the next frame from the video
        ret, frame = cap.read()

        # Break the loop if there are no more frames to read
        if not ret:
            break
        # записываем только half frames
        if frame_count % 2 == 0:
            # Generate a zero-padded filename (e.g., frame_0000.jpg)
            file_name = f"frame_{frame_count:04d}.jpg"
            file_path = os.path.join(output_folder, file_name)

            # Save the current frame as an image
            cv2.imwrite(file_path, frame)
            if (frame_count % 100 ==0):
                print(frame_count)

        frame_count += 1

    # Release resources
    cap.release()
    print(f"Done! Successfully extracted {frame_count} frames.")

```
